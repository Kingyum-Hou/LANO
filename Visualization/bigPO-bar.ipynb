```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
# 柱子数据
labels = ['A', 'B', 'C', 'D', 'E']
values = [13.8, 8.3, 9.0, 7.19]  # 第五个是用于百分比标注

colors = ['#f9f59a', '#3ec2f1', '#39e5d3', '#8aaae5']
```

```python
plt.clf()
fig, (ax_top, ax_bottom) = plt.subplots(
    2, 1, sharex=True, figsize=(7, 7),
    gridspec_kw={'height_ratios': [1, 5], 'hspace': 0.1}  # 关键：hspace 控制子图间距
)

# 设置断轴范围
ax_top.set_ylim(13.5, 14)  # 顶部图仅显示较高的值
ax_bottom.set_ylim(7, 10)  # 底部图显示正常范围

# 绘制柱子
bars_top = ax_top.bar(labels[:3], values[:3], color=colors[:3])
bars_bottom = ax_bottom.bar(labels[:3], values[:3], color=colors[:3])

# 添加第五个柱子（只有底部）
ax_bottom.bar('E', values[3], color=colors[3])

# 添加数值文本
ax_top.text(0, values[0] + 0.05, f'{values[0]:.2f}', ha='center', va='bottom', fontsize=9)
ax_bottom.text(1, values[1] + 0.1, f'{values[1]:.2f}', ha='center', va='bottom', fontsize=9)
ax_bottom.text(2, values[2] + 0.1, f'{values[2]:.2f}', ha='center', va='bottom', fontsize=9)
ax_bottom.text(3, values[3] + 0.1, f'{values[3]:.2f}', ha='center', va='bottom', fontsize=9)

# 断轴线
d = 0.02  # size of diagonal lines
ax_top.plot((-d, +d), (-d, +d), transform=ax_top.transAxes, color='k', clip_on=False)
ax_top.plot((1 - d, 1 + d), (-d, +d), transform=ax_top.transAxes, color='k', clip_on=False)

ax_bottom.plot((-d, +d), (1 - d, 1 + d), transform=ax_bottom.transAxes, color='k', clip_on=False)
ax_bottom.plot((1 - d, 1 + d), (1 - d, 1 + d), transform=ax_bottom.transAxes, color='k', clip_on=False)

# 添加虚线
ax_bottom.axhline(y=8.3, color='black', linestyle='--', linewidth=1)

# 添加红色百分比下降
ax_bottom.text(3.5, 8.3, '-13.4%', color='red', fontsize=12, fontweight='bold', va='center')

# 添加下箭头弯曲符号（简化版）
ax_bottom.annotate('', xy=(3.5, 8.3), xytext=(3.5, 7.5),
                   arrowprops=dict(arrowstyle='-|>', color='blue', lw=2, connectionstyle='arc3,rad=-0.5'))

# 去除多余边框
ax_top.spines['bottom'].set_visible(False)
ax_bottom.spines['top'].set_visible(False)
ax_top.tick_params(bottom=False, labelbottom=False)


plt.tight_layout()
plt.show()
```

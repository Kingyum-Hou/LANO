```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib
from matplotlib import rcParams
from matplotlib import rcParams, font_manager
# sudo apt install fonts-mononoki

# 指定字体文件路径（根据你的查询结果）
font_path = "/usr/share/fonts/truetype/mononoki/mononoki-Bold.ttf"
font_manager.fontManager.addfont(font_path)

# 创建字体属性
mononoki_font = font_manager.FontProperties(fname=font_path)

# 设置全局字体
rcParams['font.family'] = mononoki_font.get_name()
# 设置全局字体大小
rcParams['font.size'] = 20          # 全局默认字体大小
# 设置全局字体加粗
#rcParams['axes.labelweight'] = 'bold'     # 坐标轴标签加粗
#rcParams['axes.titleweight'] = 'bold'     # 标题加粗



# 构建数据
wout_mpt = pd.DataFrame({
    '0': [0.008, 0.0074, 0.0069],
    '0.25': [0.4837, 0.5042, 0.4834],
    '0.5': [0.6688, 0.6584, 0.6805],
}, index=[200, 500, 800])

with_mpt = pd.DataFrame({
    '0': [0.0125, 0.0091, 0.0085],
    '0.25': [0.0287, 0.0232, 0.0275],
    '0.5': [0.1051, 0.0895, 0.0934],
}, index=[200, 500, 800])

# 色阶范围
vmin = min(wout_mpt.min().min(), with_mpt.min().min())
vmax = max(wout_mpt.max().max(), with_mpt.max().max())

# 创建图和子图（不使用 sharey，以方便微调）
fig, axes = plt.subplots(1, 2, dpi=600, figsize=(10, 5.5))

# 画两个热力图
sns_plot1 = sns.heatmap(wout_mpt, annot=True, fmt=".4f", cmap="YlGnBu", vmin=vmin, vmax=vmax, ax=axes[0], cbar=False)
axes[0].set_title("w/o MPT")
axes[0].set_xlabel("Missing Rate")
axes[0].set_ylabel("Data Size")

sns_plot2 = sns.heatmap(with_mpt, annot=True, fmt=".4f", cmap="YlGnBu", vmin=vmin, vmax=vmax, ax=axes[1], cbar=False)
axes[1].set_title("with MPT")
axes[1].set_xlabel("Missing Rate")
axes[1].set_ylabel("")

# 添加共享 colorbar（右侧中间）
from mpl_toolkits.axes_grid1 import make_axes_locatable

# 在第二个子图旁边分配 colorbar 区域
divider = make_axes_locatable(axes[1])
cax = divider.append_axes("right", size="5%", pad=0.05)

# 用第二个图的 mappable 创建 colorbar
fig.colorbar(sns_plot2.collections[0], cax=cax, label="Relative L2 Error")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib
from matplotlib import rcParams
from matplotlib import rcParams, font_manager
# sudo apt install fonts-mononoki

# 指定字体文件路径（根据你的查询结果）
font_path = "/usr/share/fonts/truetype/mononoki/mononoki-Bold.ttf"
font_manager.fontManager.addfont(font_path)

# 创建字体属性
mononoki_font = font_manager.FontProperties(fname=font_path)

# 设置全局字体
rcParams['font.family'] = mononoki_font.get_name()
# 设置全局字体大小
rcParams['font.size'] = 13          # 全局默认字体大小

# 数据
x1 = [100, 250, 500, 800]
y1 = [0.4056, 0.2728, 0.2223, 0.1634]

x2 = [4, 8, 12, 16]
y2 = [0.186, 0.1435, 0.1244, 0.1185]

# 创建画布
fig, axes = plt.subplots(1, 2, figsize=(7, 3), dpi=600)

# ---------------- 图 (a) ----------------
axes[0].plot(x1, y1, '-o', color='red', linewidth=3)
axes[0].set_xlabel('#training cases')
axes[0].set_ylabel('Relative L2')
axes[0].set_title('(a) Data')
# 只显示这几个 x 值
axes[0].set_xticks([100, 250, 500, 800])

# 只显示这几个 y 值（可选）
axes[0].set_yticks([0.15, 0.20, 0.25, 0.30, 0.40])

# ---------------- 图 (b) ----------------
axes[1].plot(x2, y2, '-o', color='red', linewidth=3)
axes[1].set_xlabel('#layers')
axes[1].set_title('(b) Parameters')
# 只显示这几个 x 值
axes[1].set_xticks([4, 8, 12, 16])

# 只显示这几个 y 值（可选）
axes[1].set_yticks([0.10, 0.12, 0.15, 0.20])

# 调整布局
plt.tight_layout()
plt.show()
```
